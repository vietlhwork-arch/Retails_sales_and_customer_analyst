```python
# ==========================================
# 2. GỌI XGBOOST ĐỂ GẮN NHÃN DỰ BÁO TRẢ HÀNG
# ==========================================
print("--- Đang dự báo khả năng trả hàng ---")

# 1. Load mô hình
xgb_model = joblib.load('../app/xgboost_returns_model.pkl')

# 2. Load dữ liệu AI (để dự đoán) và Dữ liệu Gốc (để lấy Key)
df_xgb = pd.read_csv('../data/processed/ml_features_xgboost.csv')

# Đã sửa lại đúng tên file đang có trong máy của bạn:
df_raw = pd.read_csv('../data/processed/master_sales.csv') 

# 3. Yêu cầu AI gắn nhãn
features = [col for col in df_xgb.columns if col != 'is_returned']
X_predict = df_xgb[features] 

# Tạo một bảng dữ liệu MỚI dành riêng cho Power BI
df_powerbi = pd.DataFrame()

# 4. Gắn lại các Key quan trọng từ file gốc sang bảng mới
# LƯU Ý: Dựa vào lịch sử ảnh chụp của bạn, các cột này đang viết thường. 
# Nếu báo lỗi KeyError, bạn hãy mở file master_sales.csv ra xem nó viết Hoa hay thường để chỉnh lại nhé!
df_powerbi['customerkey'] = df_raw['customerkey']
df_powerbi['productkey'] = df_raw['productkey']
df_powerbi['OrderDate'] = df_raw['orderdate'] 

# 5. Gắn kết quả dự báo của AI vào bảng mới
df_powerbi['Predicted_Return'] = xgb_model.predict(X_predict) 
df_powerbi['Return_Probability'] = xgb_model.predict_proba(X_predict)[:, 1] 

# 6. Xuất file CSV số 2
xgb_output_path = '../data/processed/PowerBI_Returns_Forecast.csv'
df_powerbi.to_csv(xgb_output_path, index=False)
print(f"✅ Đã xuất bảng dự báo hoàn trả CÓ CHỨA KEY: {xgb_output_path}")
```
