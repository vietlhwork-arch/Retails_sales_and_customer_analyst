# 01 - Data Integration
Load raw CSVs, standardize columns, and build master_sales.

```python
# ============================================================
# 01 - DATA INTEGRATION
# AdventureWorks Analytics
# ============================================================

# ============================================================
# IMPORTS
# ============================================================

import sys
from pathlib import Path

import pandas as pd

# ============================================================
# PROJECT ROOT SETUP
# ============================================================

PROJECT_ROOT = Path.cwd().parent.resolve()

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("=" * 60)
print("PROJECT ROOT")
print(PROJECT_ROOT)
print("=" * 60)

# ============================================================
# CHECK STRUCTURE
# ============================================================

SRC_PATH = PROJECT_ROOT / "src"

print("SRC EXISTS:", SRC_PATH.exists())

if SRC_PATH.exists():

    print("\nFILES IN SRC:")

    for file in SRC_PATH.glob("*.py"):
        print(" -", file.name)

# ============================================================
# IMPORT ETL MODULE
# ============================================================

from src.data_preprocessing import main as preprocess_main

print("\n✅ data_preprocessing imported successfully")

# ============================================================
# DEFINE PATHS
# ============================================================

RAW_DIR = PROJECT_ROOT / "data" / "raw"

PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"

PROCESSED_DIR.mkdir(
    parents=True,
    exist_ok=True
)

print("\nRAW DIRECTORY")
print(RAW_DIR)

print("\nPROCESSED DIRECTORY")
print(PROCESSED_DIR)

# ============================================================
# RUN ETL
# ============================================================

print("\n🚀 RUNNING ETL PIPELINE...\n")

preprocess_main(
    str(RAW_DIR),
    str(PROCESSED_DIR)
)

print("\n✅ ETL COMPLETED")

# ============================================================
# VERIFY OUTPUT
# ============================================================

MASTER_FILE = (
    PROCESSED_DIR
    / "master_sales.csv"
)

print("\nMASTER FILE EXISTS:", MASTER_FILE.exists())

MASTER_FILE

# ============================================================
# LOAD RESULT
# ============================================================

master = pd.read_csv(MASTER_FILE)

print("\nMASTER SHAPE")
print(master.shape)

master.head()

# ============================================================
# DATA INFO
# ============================================================

master.info()

# ============================================================
# MISSING VALUES REPORT
# ============================================================

missing_report = pd.DataFrame({
    "column": master.columns,
    "missing": master.isna().sum().values,
    "missing_pct": (
        master.isna().mean() * 100
    ).round(2).values
})

missing_report.sort_values(
    "missing_pct",
    ascending=False
).head(20)

# ============================================================
# NUMERIC SUMMARY
# ============================================================

master.describe()

# ============================================================
# BUSINESS METRICS
# ============================================================

if "revenue" in master.columns:

    print(
        f"\nTotal Revenue : "
        f"{master['revenue'].sum():,.2f}"
    )

if "profit" in master.columns:

    print(
        f"Total Profit  : "
        f"{master['profit'].sum():,.2f}"
    )

if "orderquantity" in master.columns:

    print(
        f"Total Quantity: "
        f"{master['orderquantity'].sum():,.0f}"
    )

# ============================================================
# DATA DICTIONARY
# ============================================================

REPORTS_DIR = PROJECT_ROOT / "reports"

REPORTS_DIR.mkdir(
    parents=True,
    exist_ok=True
)

dictionary = pd.DataFrame({
    "column": master.columns,
    "dtype": master.dtypes.astype(str)
})

dictionary.to_csv(
    REPORTS_DIR / "data_dictionary.csv",
    index=False
)

print(
    "\n✅ Data dictionary saved to:"
)

print(
    REPORTS_DIR / "data_dictionary.csv"
)

# ============================================================
# SAVE SAMPLE
# ============================================================

master.head(100).to_csv(
    REPORTS_DIR / "master_sales_sample.csv",
    index=False
)

print(
    "\n✅ Sample file saved"
)

# ============================================================
# DONE
# ============================================================

print("\n🎉 DATA INTEGRATION NOTEBOOK FINISHED")
```
